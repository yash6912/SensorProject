INSIGHT
Pretty good amount of them(either first 50 of random 50) either are constant (have 0 standard deviation) or have left skewness and right skewness. it ain't possible to analyze each feature and deal with its outliers individually thus we outta depend upon the scaling

for the feature with 0 standard deviation we can straight away drop them and for others that do have outliers ,we gotta go ahead with the robust scaling

```python
## get columns to drop from the dataset
## will drop the columns with zero standard deviation as they are not influencing the target variable in any way

def get_cols_with_zero_std(df : pd.DataFrame):
    # Returns a list of columns with zero standard deviation
    cols_to_drop = []
    num_cols = [col for col in df.columns if df[col].dtype != 'O'] # numerical columns only
    for col in df.columns:
        if df[col].std() == 0:
            cols_to_drop.append(col)
    return cols_to_drop


def get_redundant_cols(df : pd.DataFrame, missing_threshold : float = 0.9):
    # Returns a list of redundant columns having missing values more than the certain threshold
    col_missing_ratios = df.isna().sum().div(df.shape[0])
    cols_to_drop = list(col_missing_ratios[col_missing_ratios > missing_threshold].index)
    return cols_to_drop
```
